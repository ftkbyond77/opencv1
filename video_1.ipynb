```python
import cv2
import numpy as np
```

```python
video_path = 'Nvidia CEO Jensen Huang_ People with really high expectations have very low resilience.mp4'  # Replace with your video file
```

# Processing Original

```python
cap = cv2.VideoCapture(video_path)
width  = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps    = cap.get(cv2.CAP_PROP_FPS)
```

```python
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
```

```python
out = cv2.VideoWriter('ai_resume_original.mp4', fourcc, fps, (width, height))

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # Font for text
    font = cv2.FONT_HERSHEY_SIMPLEX

    # Add 'Processing...' and resume info in top-right corner
    text = 'Processing...'
    position = (width - 250, 50)
    cv2.putText(frame, text, position, font, 0.7, (255, 255, 255), 2)

    # Add resume info
    cv2.putText(frame, 'Name: Jensen Huang_', (width - 250, 100), font, 0.6, (255, 255, 255), 1)
    cv2.putText(frame, 'Position: Nvidia CEO', (width - 250, 150), font, 0.6, (255, 255, 255), 1)
    cv2.putText(frame, 'Skills: Ai GPU', (width - 250, 200), font, 0.6, (255, 255, 255), 1)

    # Write frame to output
    out.write(frame)

cap.release()
out.release()
```

# Processing Gradient Detection

```python
out = cv2.VideoWriter('gradient_output.mp4', fourcc, fps, (width, height))

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # Convert frame to grayscale
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Apply Sobel gradient detection
    sobelx = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
    sobely = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)
    sobel_combined = cv2.magnitude(sobelx, sobely)
    sobel_combined = np.uint8(np.clip(sobel_combined, 0, 255))

    # Font for text
    font = cv2.FONT_HERSHEY_SIMPLEX

    # Add 'Processing...' and resume info
    cv2.putText(sobel_combined, 'Processing...', (width - 250, 50), font, 0.7, 255, 2)
    cv2.putText(sobel_combined, 'Name: Jensen Huang_', (width - 250, 100), font, 0.6, 255, 1)
    cv2.putText(sobel_combined, 'Position: Nvidia CEO', (width - 250, 150), font, 0.6, 255, 1)
    cv2.putText(sobel_combined, 'Skills: Ai GPU', (width - 250, 200), font, 0.6, 255, 1)

    # Convert to 3-channel
    sobel_bgr = cv2.cvtColor(sobel_combined, cv2.COLOR_GRAY2BGR)
    out.write(sobel_bgr)

cap.release()
out.release()
```

# Processing Morphological

```python
out = cv2.VideoWriter('morph_output.mp4', fourcc, fps, (width, height))

# Define a kernel for morphological operations
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # Convert frame to grayscale
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Apply binary threshold to prepare for morphology
    _, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

    # Apply morphological opening to remove noise
    morph = cv2.morphologyEx(thresh, cv2.MORPH_OPEN, kernel)

    # Font for text
    font = cv2.FONT_HERSHEY_SIMPLEX
    cv2.putText(morph, 'Processing...', (width - 250, 50), font, 0.7, 255, 2)
    cv2.putText(morph, 'Name: Jensen Huang_', (width - 250, 100), font, 0.6, 255, 1)
    cv2.putText(morph, 'Position: Nvidia CEO', (width - 250, 150), font, 0.6, 255, 1)
    cv2.putText(morph, 'Skills: Ai GPU', (width - 250, 200), font, 0.6, 255, 1)

    # Convert to 3-channel for output compatibility
    morph_bgr = cv2.cvtColor(morph, cv2.COLOR_GRAY2BGR)
    out.write(morph_bgr)

cap.release()
out.release()
```

# Full

```python
cap = cv2.VideoCapture(video_path)

# Get video properties
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)
frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
duration = frame_count / fps

# Prepare output
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter('Ai_Resume_opencv.mp4', fourcc, fps, (width, height))

# Define stages
stages = ['Original', 'Grayscale', 'Gradient Detection', 'Morphological']
stage_duration = duration / len(stages)
stage_frame_count = int(stage_duration * fps)

# Define kernel for morphology
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))

# Font for text
font = cv2.FONT_HERSHEY_SIMPLEX

frame_index = 0

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    current_stage_index = min(frame_index // stage_frame_count, len(stages) - 1)
    stage = stages[current_stage_index]

    if stage == 'Original':
        processed_frame = frame.copy()

    elif stage == 'Grayscale':
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        processed_frame = cv2.cvtColor(gray, cv2.COLOR_GRAY2BGR)

    elif stage == 'Gradient Detection':
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        grad_x = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
        grad_y = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)
        grad = cv2.convertScaleAbs(cv2.magnitude(grad_x, grad_y))
        processed_frame = cv2.cvtColor(grad, cv2.COLOR_GRAY2BGR)

    elif stage == 'Morphological':
        # Your specific morphological process
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        _, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)
        morph = cv2.morphologyEx(thresh, cv2.MORPH_OPEN, kernel)
        processed_frame = cv2.cvtColor(morph, cv2.COLOR_GRAY2BGR)

    # Add overlay text
    cv2.putText(processed_frame, 'Processing...', (width - 250, 50), font, 0.7, (255, 255, 255), 2)
    cv2.putText(processed_frame, 'Name: Jensen Huang_', (width - 250, 100), font, 0.6, (255, 255, 255), 1)
    cv2.putText(processed_frame, 'Position: Nvidia CEO', (width - 250, 150), font, 0.6, (255, 255, 255), 1)
    cv2.putText(processed_frame, 'Skills: Ai GPU', (width - 250, 200), font, 0.6, (255, 255, 255), 1)

    # Add stage label in top-left
    cv2.putText(processed_frame, f"Stage: {stage}", (50, 50), font, 1, (0, 255, 0), 2)

    # Write to output
    out.write(processed_frame)
    frame_index += 1

cap.release()
out.release()
print("Final video saved as 'processed_output.mp4'")
```
