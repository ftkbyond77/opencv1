```python
from moviepy.editor import VideoFileClip, AudioFileClip, concatenate, AudioClip
```

```python
ceo_clip = VideoFileClip('Ai_Resume_opencv.mp4')
ceo_audio = AudioFileClip('CEO sound.mp3')

ceo_clip.duration, ceo_audio.duration
```

```python
try:
    ceo_audio.duration == ceo_clip.duration
    # Set audio to video if the durations match
    final_video = ceo_clip.set_audio(ceo_audio)

    # Export final video with audio
    final_video.write_videofile("Ai_Resume_Audio.mp4", codec='libx264', audio_codec='aac')
except:
    print("Audio and video durations do not match. Export aborted.")
```

# Audio Loop

```python
from moviepy.editor import concatenate_audioclips
```

```python
try:
    # Check if audio duration is less than video duration
    if audio_clip.duration < video_clip.duration:
        # Loop the audio to match the video duration
        num_loops = int(video_clip.duration // audio_clip.duration)  # Number of full loops
        remaining_duration = video_clip.duration % audio_clip.duration  # Remaining time to loop

        # Create a looped audio clip
        looped_audio = concatenate_audioclips([audio_clip] * num_loops + [audio_clip.subclip(0, remaining_duration)])

        # Set the looped audio to the video
        final_video = video_clip.set_audio(looped_audio)

    else:
        # If the durations match, just set the audio
        final_video = video_clip.set_audio(audio_clip)

    # Export the final video with audio
    final_video.write_videofile("audio_loop.mp4", codec='libx264', audio_codec='aac')

except Exception as e:
    print(f"Audio and video durations do not match. Export aborted. Error: {e}")
```
